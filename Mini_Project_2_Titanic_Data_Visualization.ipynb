# Mini Project 2: Titanic Data Visualization
In this project, we will explore the Titanic dataset using **Pandas, Seaborn, and Matplotlib**.
We will perform basic statistics and create visualizations to understand survival patterns.

```

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset from seaborn
df = sns.load_dataset("titanic")

# Show first 5 rows
df.head()
```

## Basic Statistics

```

# Shape of dataset
print("Shape of dataset:", df.shape)

# Data types and null counts
print("\nInfo:")
print(df.info())

# Summary statistics for numeric columns
df.describe()
```

## Age Distribution

```

plt.figure(figsize=(8,5))
sns.histplot(df["age"], bins=30, kde=True)
plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

## Survival Count by Gender

```

plt.figure(figsize=(6,4))
sns.countplot(x="sex", hue="survived", data=df)
plt.title("Survival Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()
```

## Survival by Passenger Class

```

plt.figure(figsize=(6,4))
sns.countplot(x="pclass", hue="survived", data=df, palette="Set2")
plt.title("Survival Count by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()
```

## Correlation Heatmap

```

plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap of Titanic Dataset")
plt.show()
```
